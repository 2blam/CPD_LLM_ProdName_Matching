### IDEA
- Using External Data, create a vector store
- Using LLM - providing the (1) NAME, (2) OCS_NAME and (3) LONG_NAME, ask for the matched PRODUCT_NAME
- Note: if there is no best match, return NULL


```python
from sentence_transformers import SentenceTransformer
import chromadb
```

```python
embedding_model_name = 'sentence-transformers/all-MiniLM-L6-v2'
embedding_model = SentenceTransformer(embedding_model_name)
embedding_model.to('cuda')
```

### Create ChromaDB (Vector Store)
- Collection Name: product_name
- Distance Measure: Cosine Similarity, [0, 1]
    - the smaller the value, the more the similarity

```python
chroma_client = chromadb.Client()

collection = chroma_client.create_collection(
    name='product_name',
    metadata={'hnsw:space': 'cosine'}
)
```

### Prepare List of Product Data
- For each product in external dataframe
  - merge NAME, OCS_NAME and LONG_NAME as a single row (ignore NA)


```python
lst_product_data = []
for idx, row in df_internal.iterrows():
    row = row.fillna('')
    product = ';'.join([e for e in [row.NAME, row.OCS_NAME, row.LONG_NAME] if e != ''])
    lst_product_data.append(product)
```

#### Create Embeddings

```python
embeddings = []
batch_size = 32

for i in range(0, len(lst_product_data),batch_size):
    if i%100 ==0: print(i)
    batch = lst_product_data[i:i+batch_size]
    batch_embeddings = embedding_model.encode(
        batch,
        convert_to_tensor=True,
        show_progress_bar=False
    )
    embeddings.extend(batch_embeddings.cpu().numpy().tolist())
```

#### Store in ChromaDB

- ids - use to reference the internal dataframe index id
- meta - using LONG_NAME as well, ease for getting the result

```python
lst_meta = []

for n in df_internal.LONG_NAME.tolist():
    d = {'LONG_NAME': n}
    lst_meta.append(d)
```

```python
# store in chromaDB
for i in range(0, len(lst_product_data),batch_size):
    embeddings_batch = embeddings[i:i+batch_size]
    lst_product_data_batch = lst_product_data[i:i+batch_size]
    ids = [f'{j}' for j in range(i, i+batch_size)]
    lst_meta_batch = lst_meta[i:i+batch_size]
    collection.add(
        embeddings=embeddings_batch,
        documents=lst_product_data_batch,
        ids=ids,
        metadatas=lst_meta_batch
    )
```

### Search Best Product Name
- if the cosine distance >= 0.3, return None

```python
lst_match_result = []

for idx, row in df_external.iterrows():
    query_embedding = embedding_model.encode(
        row.PRODUCT_NAME,
        convert_to_tensor=True,
        show_progression_bar=False
    ).cpu().numpy().tolist()

    # query chromadb
    result = collection.query(
        query_embeddings=[query_embedding],
        n_results = 1
    )
    distance = result['distances'][0][0]
    dataframe_idx = int(result['ids'][0][0])

    if distance < 0.3: #
        #match_name = df_internal.iloc[dataframe_idx].LONG_NAME # referening ids
        match_name = result['metadatas'][0][0]['LONG_NAME']
        lst_match_result.append(match_name)
    else:# cannot find
        lst_match_result.append(None)

```

```python
df_result = df_external.copy()
df_result['internal_product_name'] = lst_match_result
```

### Prompt Approach

```python
def search_product_name(prompt, threshold=0.3):

    prompt_embedding = embedding_model.encode(prompt).tolist()

    query_kwargs = {
        "query_embeddings": [prompt_embedding],
        "n_results": 1,
    }

    results = collection.query(**query_kwargs)

    # Parse results
    dist, dataframe_idx = results["distances"][0][0], int(results["ids"][0][0])
    if dist < threshold:
        match_product_name = df_internal.iloc[dataframe_idx].LONG_NAME
        return match_product_name
    else:
        return None
```

```python
def build_standard_prompt(product_name):
    base_prompt = f"""
                    Find product matching: '{product_name}'
                    """
    return base_prompt.strip()


# example
prompt = build_standard_prompt('5 Hour Energy Grape (1.93oz)')
search_product_name(prompt)
```

```python
lst_result = []
for idx, row in df_external.iterrows():
    prompt = build_standard_prompt(row['PRODUCT_NAME'])
    result = search_product_name(prompt)
    print(prompt, '; Match:', result )
    lst_result.append(result)
```

```python
df_result_from_prompt = df_external.copy()
df_result_from_prompt['internal'] = lst_result
```
